# Nutrition Coach on Colab

1. Runtime → Change runtime type → **T4 GPU**.
2. In Secrets (key icon), add `NGROK_AUTH_TOKEN` and, optionally, `FDC_API_KEY`, with notebook access on.
3. Set `REPO_URL` below, then run both cells.

```python
import os
from google.colab import userdata

REPO_URL = "https://github.com/YOUR_GITHUB_USERNAME/NutritionCoach.git"  # change this
MODEL_ID = "google/gemma-4-E2B-it"  # or "google/gemma-4-E4B-it"

!rm -rf NutritionCoach && git clone -q $REPO_URL NutritionCoach
%cd NutritionCoach
!pip install -q -r requirements.txt

os.environ["MODEL_ID"] = MODEL_ID
os.environ["NGROK_AUTH_TOKEN"] = userdata.get("NGROK_AUTH_TOKEN")
try:
    os.environ["FDC_API_KEY"] = userdata.get("FDC_API_KEY")
except Exception:
    print("No FDC_API_KEY secret: using the rate-limited USDA DEMO_KEY")
print("Setup complete")
```

```python
import subprocess
import time
from pyngrok import ngrok

ngrok.set_auth_token(os.environ["NGROK_AUTH_TOKEN"])
server = subprocess.Popen(["python", "app.py"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
time.sleep(3)

public_url = ngrok.connect(5000)
print(f"Nutrition Coach is live at: {public_url}")
print("The first question takes a few minutes while the model downloads and loads.\n")

for line in server.stdout:
    print(line.decode(), end="")
```
